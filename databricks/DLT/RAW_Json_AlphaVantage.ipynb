```python
import requests
import json
import time
from datetime import datetime, timezone

API_KEY = ""
SYMBOLS = ["AAPL", "GOOGL", "AMZN", "MSFT"]

RAW_QUOTE_PATH = "/Volumes/my_catalog/default/raw_stock_quotes"
RAW_COMPANY_PATH = "/Volumes/my_catalog/default/raw_company_info"

run_ts = datetime.now(timezone.utc).strftime("%Y%m%d_%H%M%S")
ingested_at = datetime.now(timezone.utc).isoformat()


def write_json_to_volume(path, file_name, payload):
    dbutils.fs.mkdirs(path)
    dbutils.fs.put(
        f"{path}/{file_name}",
        json.dumps(payload),
        overwrite=False
    )


for symbol in SYMBOLS:
    quote_url = f"https://www.alphavantage.co/query?function=GLOBAL_QUOTE&symbol={symbol}&apikey={API_KEY}"
    response = requests.get(quote_url, timeout=10)
    response.raise_for_status()

    payload = {
        "symbol": symbol,
        "ingested_at": ingested_at,
        "raw_response": response.json()
    }

    write_json_to_volume(
        RAW_QUOTE_PATH,
        f"{symbol}_quote_{run_ts}.json",
        payload
    )

    time.sleep(15)


for symbol in SYMBOLS:
    company_url = f"https://www.alphavantage.co/query?function=OVERVIEW&symbol={symbol}&apikey={API_KEY}"
    response = requests.get(company_url, timeout=10)
    response.raise_for_status()

    payload = {
        "symbol": symbol,
        "ingested_at": ingested_at,
        "raw_response": response.json()
    }

    write_json_to_volume(
        RAW_COMPANY_PATH,
        f"{symbol}_company_{run_ts}.json",
        payload
    )

    time.sleep(15)
```
